```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import permutations

def mostrar_particiones_ordinales_w4_descendente():
    """
    Muestra las 24 particiones ordinales para w=4
    usando los valores {1,2,3,4} interpretados
    de mayor (4) a menor (1), con fontsize >= 14.
    """

    # Configuración global de tamaño de fuente
    plt.rcParams.update({
        'font.size': 14,
        'axes.titlesize': 14,
        'axes.labelsize': 14,
        'xtick.labelsize': 14,
        'ytick.labelsize': 14
    })

    w = 4

    # Generar todas las permutaciones
    permutaciones = list(permutations(range(1, w+1)))
    permutaciones.sort()

    print(f"Total de particiones ordinales para w=4: {len(permutaciones)}")

    fig = plt.figure(figsize=(15, 10))
    fig.suptitle(
        "24 particiones ordinales para w = 4",
        fontsize=18,  # Mayor que 14
        fontweight='bold'
    )

    for idx, perm in enumerate(permutaciones, 1):
        plt.subplot(6, 4, idx)

        valores = np.array(perm)

        plt.plot(range(w), valores, 'o-', linewidth=2, markersize=8)

        plt.xticks(range(w), range(1, w+1))
        plt.yticks([1,2,3,4])
        plt.ylim(4.5, 0.5)  # Mayor arriba

        plt.title(f'{perm}', fontsize=14)
        plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


# Ejecutar
mostrar_particiones_ordinales_w4_descendente()
```
