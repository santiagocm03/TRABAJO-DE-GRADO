```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import permutations

def mostrar_particiones_ordinales_w4():

    plt.rcParams.update({
        'font.size': 14,
        'axes.titlesize': 14,
        'axes.labelsize': 14,
        'xtick.labelsize': 14,
        'ytick.labelsize': 14
    })

    w = 4
    permutaciones = list(permutations(range(w)))  # ahora usamos 0,1,2,3
    permutaciones.sort()

    fig = plt.figure(figsize=(15, 10))
    fig.suptitle(
    r"24 particiones ordinales para $\boldsymbol{\omega = 4}$",
    fontsize=18,
    fontweight='bold'
)

    for idx, perm in enumerate(permutaciones, 1):
        plt.subplot(6, 4, idx)

        valores = np.zeros(w)

        # valores ordenados
        base = np.arange(1, w+1)

        # perm indica posiciones del orden creciente
        for orden, posicion in enumerate(perm):
            valores[posicion] = base[orden]

        plt.plot(range(1, w+1), valores, 'o-', linewidth=2, markersize=8)

        plt.xticks(range(1, w+1))
        plt.yticks([1,2,3,4])
        plt.ylim(0.5, 4.5)

        # Mostrar perm en notación 1-based
        titulo = tuple(p+1 for p in perm)
        plt.title(f'{titulo}')
        plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

mostrar_particiones_ordinales_w4()
```
